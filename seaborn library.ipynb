```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
print(sns.__version__)
```

```python
data = {
    "Name": ["Aman", "Riya", "Karan", "Priya", "Rohit"],
    "Math": [85, 90, 78, 92, 88],
    "Science": [80, 95, 75, 89, 84],
    "English": [78, 88, 72, 91, 86]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print(df)
```

```python
df = pd.read_csv("Students.csv")
print(df)
```

```python
tips = sns.load_dataset("tips")
print(tips.head())
```

```python
sns.set_style("darkgrid")
```

```python
sns.set_style("white")
sns.set_style("dark")
sns.set_style("whitegrid")
sns.set_style("ticks")
```

```python
sns.lineplot(x="Math", y="Science", data=df)

plt.title("Math vs Science")
plt.show()
```

```python
sns.barplot(x="Name", y="Math", data=df)

plt.title("Student Math Marks")
plt.show()
```

```python
sns.countplot(x="sex", data=tips)

plt.show()
```

```python
sns.scatterplot(x="total_bill", y="tip", data=tips)

plt.show()
```

```python
sns.relplot(
    x="total_bill",
    y="tip",
    hue="sex",
    data=tips
)

plt.show()
```

```python
sns.histplot(tips["total_bill"])

plt.show()
```

```python
sns.histplot(tips["total_bill"], kde=True)

plt.show()
```

```python
sns.kdeplot(tips["total_bill"])

plt.show()
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)

plt.show()
```

```python
sns.violinplot(x="day", y="total_bill", data=tips)

plt.show()
```

```python
sns.stripplot(x="day", y="total_bill", data=tips)

plt.show()
```

```python
sns.swarmplot(x="day", y="total_bill", data=tips)

plt.show()
```

```python
sns.pairplot(tips)

plt.show()
```

```python
corr = tips.corr(numeric_only=True)

sns.heatmap(corr, annot=True)

plt.show()
```

```python
sns.lmplot(
    x="total_bill",
    y="tip",
    hue="sex",
    data=tips
)

plt.show()
```

```python
sns.lmplot(
    x="total_bill",
    y="tip",
    hue="sex",
    data=tips
)

plt.show()
```
