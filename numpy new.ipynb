```python
 # Array stacking and splitting
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.vstack((a,b))) # vertical stacking - row wise
print(np.hstack((a,b))) #horizontal stacking - column wise
```

```python
c = np.array([[1,2,3], [4,5,6]])
print(c)
```

```python
print(np.hsplit(c,3))
```

```python
# Mathematical operation on array
```

```python
a = np.array([10,20,30])
print(a)
```

```python
print(a+10) # add
print(a-30) # subtract
print(a*2)  # multiply
print(a/10) # divide
```

```python
b = np.array([1,4,9])
print(b)
```

```python
print(np.square(b))
print(np.sqrt(b))
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a)
print(b)
```

```python
print(np.add(a,b))
```

```python
print(np.dot(a,b)) # dot product
```

```python
print(a.T) # Transpose
```

```python
# Stastical functions
```

```python
a = np.array([[1,2,3], [4,5,6]])
print(a)
```

```python
print(np.sum(a))
```

```python
print(np.mean(a))
print(np.median(a))
print(np.std(a))
```

```python
# Array comparison
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a)
print(b)
```

```python
print(a == b)
print(np.array_equal(a,b))
```
