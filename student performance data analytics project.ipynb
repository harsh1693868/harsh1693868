```python
import pandas as pd
import numpy as  np
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Aman", "Riya", "Karan", "Sneha", "Rahul"],
    "Maths": [85, 78, 92, 74, 88],
    "Science": [90, 81, 95, 70, 84],
    "English": [78, 85, 89, 80, 76]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.to_csv("student_data.csv", index=False)

print("CSV file created successfully")
```

```python
df = pd.read_csv("Student_data.csv")
print(df)
```

```python
print("Average Maths Marks:", df["Maths"].mean())
print("Average Science Marks:", df["Science"].mean())
print("Average English Marks:", df["English"].mean())
```

```python
print("Highest Maths marks:",
df["Maths"].max())
```

```python
top_student = df[df["Science"] == df["Science"].max()]

print(top_student)
```

```python
df["Total"] = df["Maths"] + df["Science"] + df["English"]

df["Percentage"] = (df["Total"] / 300) * 100

print(df)
```

```python
df.to_csv("updated_student_data.csv", index=False)

print("Updated CSV file saved")
```
