```python
abs(-10)  #10
pow(2, 3)  # 8
round(3, 456, 2) # 3.46
min(3, 7, 1)     # 1
max(3, 7, 1)     # 7
sum([1,2, 3])    # 6
```

```python
int("10")      # 10
float("3.14")  # 3.14
str(100)       # '100'
bool(0)        # false
list("abc")    # ['a', 'b', 'c']
tuple([1,2])   # (1, 2)
set([1, 1, 2]) # {1,2}
dict(a=1, b=2)  # {'a': 1, 'b': 2}
```

```python
len("python")          # 6
sorted([3,1, 2])       # [1,2,3]
reversed([1,2,3])      # iterator
enumerate(['a', 'b'])  # index + value
zip([1,2]), ['a', 'b'] # pair values
```
