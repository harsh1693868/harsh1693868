```python
add = lambda a,b: a + b
print(add(5,3)) # output: 8
```

```python
numbers = [1,2,3,4]

squared = list(map(lambda x: x * x, numbers))

print(squared) # ouput: [1,4,9,16]
```

```python
numbers = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x % 2 == 0, numbers))

print(even) # output: [2,4,6]

                   
```
