```python
import random

number = random.randint(1, 50)
guess = 0

while guess != number:
    guess = int(input("Guess a number between 1 and 50: "))
    
    if guess < number:
        print("Too low!")
    elif guess > number:
        print("Too high!")
    else:
        print("Correct! You guessed it.")
```
