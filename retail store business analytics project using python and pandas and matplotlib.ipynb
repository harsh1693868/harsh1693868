```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Order_ID": [101, 102, 103, 104, 105, 106],
    "Product": ["Laptop", "Phone", "Tablet", "Printer", "Monitor", "Keyboard"],
    "Category": ["Electronics", "Electronics", "Electronics", "Office", "Office", "Accessories"],
    "Price": [60000, 25000, 30000, 15000, 12000, 2000],
    "Quantity": [2, 5, 3, 2, 4, 10],
    "Month": ["Jan", "Jan", "Feb", "Feb", "Mar", "Mar"]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.to_csv("business_data.csv", index=False)

print("CSV File Created")
```

```python
df = pd.read_csv("business_data.csv")

print(df.head())
```

```python
df = pd .read_csv("Business_data.csv")
print(df.head())
```

```python
df["Revenue"] = df["Price"]
df["Quantity"]

print(df)
```

```python
total_revenue = df["Revenue"].sum()
print("Total Revenue:", total_revenue)
```

```python
best_product = df.groupby("Product")["Quantity"].sum().idxmax()

print("Best Selling Product:", best_product)
```

```python
category_sales = df.groupby("Category")["Revenue"].sum()

print(category_sales)
```

```python
monthly_sales = df.groupby("Month")["Revenue"].sum()

print(monthly_sales)
```

```python
plt.bar(df["Product"], df["Revenue"])

plt.xlabel("Products")
plt.ylabel("Revenue")
plt.title("Revenue by Product")

plt.show()
```

```python
df.to_csv("updated_business_data.csv", index=False)

print("Updated CSV Saved")
```
