```python
# Student marks analyzer using list

marks = [78, 85, 66, 78, 77]

print("Marks", marks)

# Add new mark
marks.append(88)

# Average
average = sum(marks) / len(marks)
print("Average marks:", average)

# Highest and lowest
print("Highest marks:", max(marks))
print("Lowest marks:", min(marks))

# sorted list
print("Sorted marks:", sorted(marks))
```
