```python
import matplotlib.pyplot as plt

# pyplot is a collection of functions that make matplotlib work as MatLab
```

```python
# 1 Create first chart
```

```python
# data
x = [1,2,3,4,5]
y = [10,20,25,35, 45]

plt.plot(x,y) # Line chart
plt.show()    # display chart in jp nb
```

```python
# 2 Customize chart
```

```python
plt.plot(x,y, color= 'blue', marker='o', linestyle= 'dashed', linewidth=2, markersize=12)

plt.title("Kuch bhi title hai")
plt.xlabel("x-axis label hai")
plt.ylabel("y-axis label hai")
plt.show()
```

```python
# 3 Advanced line charts - Multiple line & legends
```

```python
# data
x = [1,2,3,4,5]
y1 = [10,20,25,35, 45]
y2 = [20,30,35,45, 55]

plt.plot(x, y1, label='sales 2024') # plotting y1 data
plt.plot(x, y2, label='sales 2025') # plotting y2 data

plt.title("YoY Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend()
plt.show()
```

```python
# 4 Bar chart
```

```python
# data
x = ['A', 'B', 'C', 'D','E']
y = [10,20,25,35, 45]

plt.bar(x,y)
plt.title('Bar chart Example')
plt.show()
```

```python
# 5 Histogram
```

```python
# data
import random
data = [random.randint(1,20) for _ in range(500)]

plt.hist(data, bins=20)
plt.title("Histogram Example")
plt.show()
```

```python
# 6 Pie chart
```

```python
# used to show part-to-whole relationships/data

# data
categories = ['A', 'B', 'C', 'D','E']
sales = [10,20,25,35, 45]

plt.pie(sales, labels=categories,autopct = '%1.1f%%')
plt.title("pie chart Example")
plt.show()
```

```python
# 7 Scatter plot
```

```python
# used to find relation btwn variables

# data
y1 = [10,20,25,35, 45]
y2 = [20,30,35,45, 55]

plt.scatter(y1, y2)
plt.title("Scatter Plot Example")
plt.show()
```

```python
# 8 subplots
```

```python
# Used to show multiple charts in one figure

# data-1 - bar chart
categories = ['Mon', 'Tue', 'Wed', 'Thu','Fri']
sales = [10,20,25,35, 45]

# data-2 - Scatter plot
y1 = [10,20,25,35, 45]
y2 = [20,30,35,45, 55]

plt.figure(figsize=(10,6))

# First plot- bar chart
plt.subplot(1,2,1) # row, column, position
plt.bar(categories, sales)
plt.title("Daily sales")
plt.xlabel("Week Days")
plt.ylabel("sales")



# Second plot- Scatter chart
plt.subplot(1,2,2) # row, column, position
plt.bar(y1, y2)
plt.title("User Example")
plt.xlabel("User1")
plt.ylabel("User2")

plt.show()
```

```python
# 9 Matplotlib with pandas - real data
```

```python
# Create df
import pandas as pd

# data
data = {
    'Month' : ['jan', 'feb', 'mar', 'Apr'],
    'sales' : [12000, 11000, 13000, 25000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'], df['Sales'])
plt.title("Matplotlib with pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# 10 Save plots/ charts
```

```python
plt.bar(df['Month'], df['Sales'])
plt.title("Matplotlib with pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("Monthly_Sales_from_df.png")

plt.show()
```
