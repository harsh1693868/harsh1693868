```python
def calculate_average(marks):
    return sum(marks) / len(marks)

def find_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 60:
        return "C"
    else:
        return "Fail"

# Main
marks = []

for i in range(5):
    mark = float(input(f"Enter mark {i+1}: "))
    marks.append(mark)

average = calculate_average(marks)
grade = find_grade(average)

print("Average:", average)
print("Grade:", grade)
```
