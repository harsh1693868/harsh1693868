```python
import sys

a= [1, 2, 3]
print(sys.getrefcount(a)) # reference

b = a
print(sys.getrefcount(a)) # increases

del b
print(sys.getrefcount(a)) # decreases
```

```python
import copy

original = [[1,2], [3, 4]]
shallow = copy.copy(original)
shallow[0][0] = 99

print("original:", original)
print("shallow: ", shallow)
```
